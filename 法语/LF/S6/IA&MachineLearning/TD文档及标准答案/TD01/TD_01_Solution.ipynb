### Exercice 2 -	ETUDIANTS 

```python
import math

# Données
classe = [
    ("Paul", "Durand", "21854705", 12),
    ("Claire", "Martin", "22518750", 15),
    ("Lucas", "Bernard", "21694696", 9),
    ("Emma", "Petit", "22664784", 18),
    ("Hugo", "Moreau", "22699686", 11),
    ("Faustine", "Roux", "23085543", 14),
    ("Antoine", "Fournier", "23017691", 7),
    ("Anne", "Girard", "21611896", 16)
]

# 1. Afficher la liste complète des étudiants
print("1) Liste complète des étudiants :")
print(classe)

# 2. Afficher les informations du premier et du cinquième étudiant de la liste.
print("\n2) Premier étudiant :")
print(classe[0])
print("\n2) Cinquième étudiant :")
print(classe[4])

# 3. Afficher le prénom et la note du troisième étudiant
troisieme = classe[2]
print("\n3) Prénom et note du troisième étudiant :")
print("Prénom :", troisieme[0])
print("Note :", troisieme[3])


# 4. Calculer la moyenne des notes
notes = [etudiant[3] for etudiant in classe]
n = len(notes)

moyenne = sum(notes) / n
print("\n4) Moyenne :", round(moyenne, 2))


# 5. Calculer la médiane (sans bibliothèque externe)

notes_triees = sorted(notes)
if n % 2 == 1:
    mediane = notes_triees[n // 2]
else:
    mediane = (notes_triees[n // 2 - 1] + notes_triees[n // 2]) / 2
print("\n5) Médiane :", mediane)


# 6. Afficher le 3 meilleurs étudiants.

classe_triee_desc = sorted(classe, key=lambda e: e[3], reverse=True)
top3 = classe_triee_desc[:3]

print("\n6) Top 3 :")
for prenom, nom, num, note in top3:
    print(f"   - {prenom} {nom} ({num}) : {note}")


# 7. Calculer le taux d’admis / mention / excellent

admis = sum(1 for x in notes if x >= 10)
mention = sum(1 for x in notes if x >= 14)
excellent = sum(1 for x in notes if x >= 16)

print("\n7) Taux admis :", round(admis / n * 100, 1), "%")
print("   Taux mention (>=14) :", round(mention / n * 100, 1), "%")
print("   Taux excellent (>=16) :", round(excellent / n * 100, 1), "%")


# 8. Faire un histogramme des notes par tranches 

tranches = {
    "[0,5[": 0,
    "[5,10[": 0,
    "[10,15[": 0,
    "[15,20]": 0
}

for x in notes:
    if 0 <= x < 5:
        tranches["[0,5["] += 1
    elif 5 <= x < 10:
        tranches["[5,10["] += 1
    elif 10 <= x < 15:
        tranches["[10,15["] += 1
    elif 15 <= x <= 20:
        tranches["[15,20]"] += 1

print("\n8) Histogramme :")
for k, v in tranches.items():
    print(f"   {k} : {v}")


# 9. Rechercher un étudiant par numéro (ex: "21694696") et afficher “Admis/Ajourné”.

numero_recherche = "21694696"
trouve = False

for prenom, nom, num, note in classe:
    if num == numero_recherche:
        trouve = True
        statut = "Admis" if note >= 10 else "Ajourné"
        print(f"\n9) Recherche {numero_recherche} : {prenom} {nom} - note {note} ({statut})")
        break

if not trouve:
    print(f"\n9) Numéro {numero_recherche} introuvable.")


# 10. Vérifier que toutes les notes sont dans [0,20] (anomalies).

anomalies = []
for etudiant in classe:
    if not (0 <= etudiant[3] <= 20):
        anomalies.append(etudiant)

print("\n10) Anomalies :")
if len(anomalies) == 0:
    print("   Aucune note invalide.")
else:
    for prenom, nom, num, note in anomalies:
        print(f"   - {prenom} {nom} ({num}) : {note}")


# 11. Normaliser les notes entre 0 et 20 (min-max) et créer classe_norm.

min_note = min(notes)
max_note = max(notes)

classe_norm = []
if max_note == min_note:
    # Cas particulier : toutes les notes identiques
    for prenom, nom, num, note in classe:
        classe_norm.append((prenom, nom, num, 20.0))
else:
    for prenom, nom, num, note in classe:
        note_norm = (note - min_note) / (max_note - min_note) * 20
        note_norm = round(note_norm, 2)
        classe_norm.append((prenom, nom, num, note_norm))

print("\n11) Classe normalisée (extrait) :")
for e in classe_norm[:3]:
    print("   ", e)


# 12. Afficher un bulletin texte de la classe.

print("\n12) Bulletin de la classe")
print("-" * 35)
for prenom, nom, num, note in classe:
    statut = "Admis" if note >= 10 else "Ajourné"
    print(f"{num} - {prenom} {nom} : {note} ({statut})")

print("-" * 35)
print("Moyenne classe :", round(moyenne, 2))
print("Médiane :", mediane)

```

### Exercice 3 -	GESTION DE COMMANDES

```python
# Données
commandes = {
    "C001": ("ordinateur", 1),
    "C002": ("stylo", 10),
    "C003": ("cahier", 3),
    "C004": ("imprimante", 2),
    "C005": ("stylo", 5)
}

# 1) Fonction avec match
def prix_unitaire(produit):
    match produit:
        case "ordinateur":
            return 800
        case "imprimante":
            return 150
        case "cahier":
            return 3
        case "stylo":
            return 1
        case _:
            return 0

# 2) Fonction avec for (parcours du dictionnaire)
def afficher_factures(commandes):
    for numero, (produit, quantite) in commandes.items():
        prix = prix_unitaire(produit)
        total = prix * quantite
        print(f"{numero} - {produit} : {total} €")

# 3) Appel de fonction avec paramètre
afficher_factures(commandes)
```

```python
graphe = {
    "Paris": ["Lyon", "Lille", "Amiens"],
    "Lyon": ["Paris", "Marseille", "Grenoble"],
    "Lille": ["Paris", "Amiens"],
    "Marseille": ["Lyon", "Nice"],
    "Grenoble": ["Lyon", "Nice"],
    "Amiens": ["Lille", "Paris"],
    "Nice": ["Marseille", "Grenoble"]
}

def parcourir(ville, graphe, visitees):
    visitees.add(ville)

    for voisine in graphe[ville]:
        if voisine not in visitees:
            print(f"{ville} -> {voisine}")
            parcourir(voisine, graphe, visitees)

# Ensemble des villes déjà visitées
visitees = set()

# Appel de la fonction récursive
parcourir("Amiens", graphe, visitees)

```
