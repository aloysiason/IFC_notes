On considère une EDO linéaire d'ordre 2 avec condition initiale et finale sur $I=]0,1[$

\begin{equation}\label{Laplacien}
(1) \left\{
    \begin{array}{ll}
       y''(t)=y(t)\text{ pour tout }t\in I\\
       y(0)=0,\ \ y(1)=1
    \end{array}
\right.
\end{equation}

Supposons que l'on veuille approcher cette solution. Soit $N\geq 1$, on décompose $[0,1]$ en $N$ intervalles et on pose $t_n=\frac{n}{N}$ pour $n$ entre $0$ et $N$. On pose $h=\frac{1}{N}$ la longueur d'un intervalle. On approche $y''(t_n)$ par la différence finie suivante 

\begin{equation*}
y''(t_n)\simeq\frac{y'(t_n)-y'(t_{n-1})}{h}\simeq\frac{\frac{y_{n+1}-y_{n}}{h}-\frac{y_{n}-y_{n-1}}{h}}{h}=\frac{y_{n+1}-2y_{n}+y_{n-1}}{h^2},
\end{equation*}

ce qui nous donne le schéma

\begin{equation*}
y_0=0,\  y_N=1,\ \forall n=1,...,N-1, y_{n+1}-(2+h^2)y_{n}+y_{n-1}=0,
\end{equation*}

qu'on peut alors écrire sous forme matricielle

\begin{equation*}
\underbrace{
\begin{pmatrix}
 -(2+h^2) &  1 & 0 & ... &0 \\
 1 &  -(2+h^2) &  1 & ... &0 \\
  &  & ... &  & \\
0 & ... &  1 &  -(2+h^2) &  1\\
 0 & ... & 0 &  1 &  -(2+h^2)
\end{pmatrix}
}_{A}
\underbrace{
\begin{pmatrix}
 y_1\\
  y_2\\
 \vdots\\
 y_{N-2}\\
 y_{N-1}
\end{pmatrix}}_{Y}=\underbrace{\begin{pmatrix}
 0\\
 0\\
 \vdots\\
 0\\
  -1
\end{pmatrix}.
}_{B}
\end{equation*}

### 3. Conditionnement

Le conditionnement d'une matrice peut avoir un fort impact sur la résolution d'un système linéaire de la forme $MX=b$ et la stabilité d'une solution à des petites perturbations. On rappelle (TD2 Exercice 3) que si $x$ et $x + \delta x$ sont solutions de 

\begin{equation*}
\begin{split}
M x &= b \\
M (x+ \delta x) &= b +\delta b 
\end{split}
\end{equation*}

Alors

\begin{equation*}
\frac{\left\| \delta x \right\|_p}{\left\|x \right\|_p} \leq \mathrm{cond}_p(M) \frac{\left\|  \delta b \right\|_p}{\left\|b\right\|_p}
\end{equation*}
