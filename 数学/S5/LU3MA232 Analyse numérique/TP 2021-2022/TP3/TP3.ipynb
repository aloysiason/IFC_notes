## I) Rappels de la méthode
On rappelle le principe de la méthode de Newton. Soit une fonction $f$ dérivable sur un intervalle $I$ et s'annulant en un point $\alpha$ a priori inconnu. Pour $x_0 \in I$, on construit la suite $(x_k)_{k\in\mathbf{N}}$ par récurrence :
$$ x_{k+1} = x_k - \frac{f(x_k)}{f'(x_k)}. $$
En pratique, il faut déterminer un critère d'arrêt de l'algorithme. Par exemple, on arrête les calculs lorsque l'on trouve $x_k$ tel que $f(x_k)$ est proche de $0$ à $10^{-6}$ près. On pourra également gérer le cas éventuel où l'on a $f'(x_k) = 0$ ou $x_k$ hors de notre intervalle de travail.
