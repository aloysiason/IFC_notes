### 1.- Regex sans caractères spéciaux

```python
import re #导入 Python 的re模块（regular expression 的缩写）

#定义一个字符串变量txt，存储需要被检索的文本内容。末尾的反斜杠\是 Python 的换行续行符.
txt = "L’IA transforme le monde moderne. L’IA est utilisée en médecine,\
en finance et dans l’éducation. Certains étudiants veulent se spécialiser en IA."

regex ="IA"
res = re.search(regex, txt) #调用re.search()函数，在txt字符串中从左到右查找第一个匹配regex模式的子串。

if res :
   print(f"La première occurrence du motif '{regex}' a été  \
détectée entre l'indice {res.start()} et l'indice {res.end()-1}")
else :
   print(f"Le motif '{regex}' est absent dans '{txt}'")
```

```python
import re
#"""可用于引入长文本"""
texte = """L’IA transforme le monde moderne.
De nombreuses entreprises investissent dans l’IA.
L’IA est utilisée en médecine, en finance et dans l’éducation.
Certains étudiants veulent se spécialiser en IA.
L’IA évolue très rapidement."""

# Regex pour trouver le mot exact "IA"
pattern = r"\bIA\b" #r表示原始字符串，\b是正则中的 “单词边界”，确保匹配的是独立的单词 IA
occurrences = re.findall(pattern, texte) ##这个函数会遍历texte，找出所有符合pattern的子串

print("Nombre d'occurrences de 'IA' :", len(occurrences))
```

### 2.- Regex avec caractères spéciaux

```python
import re

def print_res(res, regex, source):
   if res :
       print(f"{regex} - '{source}' : True")
   else :
       print(f"{regex} - '{source}' : False")

txt_1 = "Un pâtissier qui pâtissait chez un tapissier qui tapissait, \
demanda un jour au tapissier qui tapissait : vaut-il mieux pâtisser \
chez un tapissier qui tapisse ou tapisser chez un pâtissier qui pâtisse ?"

#regex_1 ="pâ"
#regex_1 ="^pâ" 以 "pâ" 开头的文本
#regex_1 ="^Un" 以 "Un" 开头的文本
#regex_1 ="tapiss.." tapiss + 任意两个字符
#regex_1 ="...isse"  任意三个字符 + "isse"
regex_1 ="...isse \\?$"  # isse + 空格 + 问号+ 文章的结尾
#regex_1 ="[a-z][a-z][a-z]" 任意三个小写字母连续出现。

a1 = re.search(regex_1, txt_1)
print_res(a1, regex_1, txt_1)
```

```python
import re

txt = "YabbbYaaYbbY"
print(f"txt={txt}")

regex_1 = re.compile("a*b+")
res_1= regex_1.findall(txt)
print(f"res_1={res_1}")

regex_2 = re.compile("Ya?b")
res_2= regex_2.findall(txt)
print(f"res_2={res_2}")
```

### 4.- Regex avec un nombre d’occurrences limité

```python
import re

txt = "abaabbaaabbb"
print(f"txt={txt}")

regex_1 = re.compile("a{3}b{3}")
res_1= regex_1.findall(txt)
print(f"res_1={res_1}")


regex_2 = re.compile("a{2,3}b{2,3}")
res_2= regex_2.findall(txt)
print(f"res_2={res_2}")
```

### 5.- Regex avec groupage des résultats

```python
import re

txt = "x=48.85 et y=2.35"

regex = re.compile("([0-9]+)\\.([0-9]+)")
res = regex.findall(txt)

print(txt)
for n1,n2 in res:
   print(f"n1={n1}; n2={n2}")
```

### 6.- Mision detox

```python
import re

# Phrase initiale
texte = "J'ai mangé plein de bonbons, de chocolat, de pizza, de gâteaux et de chips."

# Regex pour remplacer tout ce qui vient après "plein de"
pattern = r"plein de .*"

# Remplacement
nouveau_texte = re.sub(pattern, "seulement une salade.", texte)

# Affichage
print("Phrase originale :")
print(texte)

print("\nPhrase modifiée :")
print(nouveau_texte)
print("\nChallenge « stop au sucre » réussi ! ;-) ")
```

### 7.- Transformation de CSV en XML

```python
import re

# Fichiers
input_file = "livres.csv"
output_file = "livres.xml"

# Regex (3 groupes) et template XML
pattern = re.compile(r"^(.*?);(.*?);(.*?)$")
replacement = r"<Livre><Auteur>\1</Auteur><Titre>\2</Titre><Editeur>\3</Editeur></Livre>"

xml_lines = []

with open(input_file, "r", encoding="utf-8") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue  # ignorer lignes vides

        m = pattern.match(line)
        if not m:
            # Si une ligne n'est pas au bon format, on l'ignore
            continue

        xml_line = pattern.sub(replacement, line)
        xml_lines.append(xml_line)

# Écriture du fichier XML final (avec une racine)
with open(output_file, "w", encoding="utf-8") as f:
    f.write("<Livres>\n")
    for x in xml_lines:
        f.write("  " + x + "\n")
    f.write("</Livres>\n")

print("Fichier XML créé :", output_file)
```
