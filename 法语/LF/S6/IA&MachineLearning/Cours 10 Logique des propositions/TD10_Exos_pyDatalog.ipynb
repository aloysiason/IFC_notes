## Exercice 1 - Questions de cours

(1) Quelle est la différence entre un fait et une règle en pyDatalog ?
事实与规则的区别是什么？

Un fait est une information déclarée comme vraie dans la base. Une règle permet de déduire une nouvelle information à partir d’autres informations.

事实是知识库中声明为真的信息；规则用于从已有信息推导出新信息。

(2) À quoi sert pyDatalog.create_terms(...) ?

pyDatalog.create_terms(...) sert à déclarer les termes que l’on va utiliser : variables, prédicats, fonctions logiques, etc.

用于声明将要使用的项：变量、谓词、逻辑函数等。

(3) Pourquoi les variables logiques commencent-elles généralement par une
majuscule ?
为什么逻辑变量通常以大写字母开头？

Les variables logiques commencent généralement par une majuscule par convention, et parce que cela permet de les distinguer des noms de prédicats ou de fonctions.

逻辑变量约定以大写开头，以便与谓词名、函数名区分。

(4) Que signifie le symbole <= lorsqu’on écrit une règle ?
编写规则时符号 <= 表示什么？

Le symbole <= se lit ici comme « est vrai si » et sert à écrire une règle.

<= 表示 “当…… 时为真”，用于编写规则。

## Exercice 6 - Agrégation simple 简单聚合
On considère :

```python
from pyDatalog import pyDatalog
pyDatalog.clear()
pyDatalog.create_terms('X, Y, suit, nb_etudiants, len_')

+suit('Alice', 'Python')
+suit('Bob', 'Python')
+suit('Claire', 'Logique')
+suit('David', 'Python')
+suit('Emma', 'Logique')
```

### 1. Définir une agrégation `nb_etudiants[Y]` qui compte le nombre d'étudiants inscrits dans chaque cours.
定义聚合函数 nb_etudiants [Y]，统计每门课的选课人数。

```python
(nb_etudiants[Y] == len_(X)) <= suit(X, Y)
```

## Exercice 7 - Compléter un petit programme
Compléter le code suivant pour exprimer :
- Alice lit ;
- Bob lit ;
- si quelqu'un lit, alors cette personne apprend.

```python
from pyDatalog import pyDatalog
pyDatalog.clear()
pyDatalog.create_terms('X, lit, apprend')

# faits
+lit('Alice')
+lit('Bob')

# regle
apprend(X) <= lit(X)

# requete
print(apprend(X))
```
