### Exercice 1 -	MISSION DETOX 

```python
import re

# Phrase initiale
texte = "J'ai mangé plein de bonbons, de chocolat, de pizza, de gâteaux et de chips."

# Regex pour remplacer tout ce qui vient après "plein de"
pattern = r"plein de .*"

# Remplacement
nouveau_texte = re.sub(pattern, "seulement une salade.", texte)

# Affichage
print("Phrase originale :")
print(texte)

print("\nPhrase modifiée :")
print(nouveau_texte)
print("\nChallenge « stop au sucre » réussi ! ;-) ")
```

### Exercice 2 -	NUMEROS DE TELEPHONE EN FRANCE

```python
import re

# Lire le fichier
with open("telephones.txt", "r") as fichier:
    numeros = fichier.readlines()

print("Numéros français trouvés :\n")

# Regex pour numéros français
pattern = r"^(\+33|0033)\d{9}$"

for numero in numeros:
    numero = numero.strip()   # supprime les espaces inutiles au début et à la fin d’une chaîne de caractères.
    
    if re.match(pattern, numero):
        print(numero)
```

### Exercice 3 -	COMPTER LES NUMEROS PAR PAYS

```python
import re

# Dictionnaire des pays
pays_codes = {
    "France": r"^(\+33|0033)",
    "Argentine": r"^\+54",
    "Royaume-Uni": r"^\+44",
    "USA": r"^\+1",
    "Allemagne": r"^\+49",
    "Espagne": r"^\+34",
    "Italie": r"^\+39"
}

# Dictionnaire pour compter
compteur = {}

# Initialisation
for pays in pays_codes:
    compteur[pays] = 0

# Lecture du fichier
with open("telephones.txt", "r") as fichier:
    numeros = fichier.readlines()

# Parcours des numéros
for numero in numeros:
    numero = numero.strip()
    
    for pays, pattern in pays_codes.items():
        if re.match(pattern, numero):
            compteur[pays] += 1
            break

# Affichage des résultats
print("Nombre de numéros par pays :\n")
for pays, nb in compteur.items():
    print(pays, ":", nb)
```

### Exercice 4 -	OPERATION “RATTRAPAGE”

```python
import re
import random

# Regex avec groupes
# pattern = r"(\w+)\s*:\s*(\d+)"
pattern = r"^(.+?)\s*:\s*(\d+)\s*$"

with open("notes.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        ligne = ligne.strip()
        m = re.match(pattern, ligne)
        if m:
            nom = m.group(1)
            note = int(m.group(2))

            if note < 10:
                rattrapage = min(20, note + random.randint(5, 10))
                print(f"Nom = {nom} | Note = {note} | Rattrapage = {rattrapage}")
            else:                
                print(f"Nom = {nom} | Note = {note} ")

```

### Exercice 5 -	TRANSFORMATION DES DATES

```python
import re

# Dictionnaire des mois 
mois = {
    "janvier": "01",
    "février": "02",
    "mars": "03",
    "avril": "04",
    "mai": "05",
    "juin": "06",
    "juillet": "07",
    "août": "08",
    "septembre": "09",
    "octobre": "10",
    "novembre": "11",
    "décembre": "12"
}

# Regex : jour de semaine + (jour) + (mois) + (année)
pattern = r"^\s*\w+\s+(\d{1,2})\s+([^\s]+)\s+(\d{4})\s*$"

with open("dates.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        ligne = ligne.strip()

        m = re.match(pattern, ligne)
        if not m:
            print("Ligne ignorée (format invalide) :", ligne)
            continue

        jour = m.group(1)          # ex: "9"
        nom_mois = m.group(2).lower()  # ex: "mai"
        annee = m.group(3)         # ex: "1913"

        if nom_mois not in mois:
            print("Mois inconnu :", nom_mois, "dans", ligne)
            continue

        jj = jour.zfill(2)
        mm = mois[nom_mois]

        print(f"{ligne} ----> {jj}/{mm}/{annee}")
```
