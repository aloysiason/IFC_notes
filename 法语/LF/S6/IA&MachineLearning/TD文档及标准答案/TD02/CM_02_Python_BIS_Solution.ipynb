### Exercice 2 – Création de smileys

```python
yeux = [":", ";", ":", ":", "x", "8"]
bouches = [")", ")", "(", "D", "D", ")"]

for indice, (oeil, bouche) in enumerate(zip(yeux, bouches)):
    smiley = oeil + bouche
    print("Smiley", indice, " -->  ", smiley)
```

### Example - Fonction SANS paramètres

```python
import random as rd 
 
def print_random_list(): 
 
   rd_list = [] 
   for i in range(5): 
       rd_list.append(rd.randint(1,10)) 
 
   print(f"rd_list={rd_list}") 
 
rd.seed(1) 
print_random_list() 
```

### Example - Fonctions AVEC paramètres

```python
import random as rd 
 
def print_random_list(n, min_val, max_val): 
 
   rd_list = [] 
   for i in range(n): 
       rd_list.append(rd.randint(min_val, max_val)) 
 
   print(f"rd_list={rd_list}") 
 
rd.seed(1) 
print_random_list(5, 1, 10) 
print_random_list(10, 1, 100) 
```

### Example - Portée des variables

```python
# Variables globales
x = 2
y = 4

def somme(x):
    global y
    y = y + x
    print("Dans somme :")
    print("1) x =", x, "     y =", y)
    print("-" * 30)

def multiplication():
    y = 5
    y = y * x
    print("Dans multiplication :")
    print("2) x =", x, "     y =", y)
    print("-" * 30)

# Appels des fonctions
multiplication()
somme(8)

# Valeurs finales
print("Après les appels :")
print("3) x =", x, "     y =", y)
```

### Example - Fonctions récursives

```python
def factorial(n): 
   print(n, " * ", end=" ")  # Par défaut, print termine par un retour à la ligne, mais ici la fin de print est remplacée par un espace.
   if n<2: 
       return 1 
   else: 
       return n*factorial(n-1) 
 
print(f"On va calculer le 4! = ", end=" ") 
print(f" = {factorial(4)}", end=" ")  
```

### Exercice 3 - Somme des n premiers nombres

```python
def somme_n(n):
    # Cas de base
    if n == 0:
        return 0
    # Cas récursif
    else:
        return n + somme_n(n - 1)

# Programme principal
n = int(input("Entrez un entier n : "))
resultat = somme_n(n)
print("La somme des", n, "premiers nombres est :", resultat)
```
