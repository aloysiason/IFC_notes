Exercise 2 - Création de smileys

```python
yeux = [':', ';', 'X','8']
bouches = ['(', ')', 'D', ')']

for indice, (y, b) in enumerate(zip(yeux, bouches)):
    smiley = y + b
    print("Smiley", indice, " -->  ", smiley)
```

Example - Portée des variables

```python
# Variables globales
x = 2
y = 4

def somme(x):
    global y
    y = y + x
    print("Dans somme :")
    print("1) x =", x, "     y =", y)
    print("-" * 30)

def multiplication():
    y = 5
    y = y * x
    print("Dans multiplication :")
    print("2) x =", x, "     y =", y)
    print("-" * 30)

# Appels des fonctions
somme(8)
multiplication()

# Valeurs finales
print("Après les appels :")
print("3) x =", x, "     y =", y)
```

Exercice3 - Somme des n premiers nombres

```python
def somme_n(n):
    # Cas de base (0 的累加和就是 0)
    if n == 0:
        return 0
    # Cas récursif (当n ≠ 0时，执行递归求和逻辑)
    else:
        return n + somme_n(n - 1)

# Programme principal
n = int(input("Entrez un entier n : "))
resultat = somme_n(n)
print("La somme des", n, "premiers nombres est :", resultat)
```
