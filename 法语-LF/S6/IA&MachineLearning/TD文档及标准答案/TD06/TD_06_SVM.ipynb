## Exercice 1 — Premier SVM sans texte

````python
from sklearn.svm import SVC

X = [
    [5, 0],
    [6, 0],
    [7, 1],
    [15, 3],
    [18, 4],
    [20, 5]
]

y = ["neutre", "neutre", "neutre", "expressif", "expressif", "expressif"]

model = SVC()
model.fit(X, y)

print(model.predict([[8, 0]]))
print(model.predict([[17, 4]]))
````

## Exercice 2 — Question ou affirmation ?

````python
from sklearn.svm import SVC

#目标：基于两个语言学特征对句子进行分类
#每个句子用两个特征编码：是否有问号/疑问词：0 或 1

X = [
    [1, 1],
    [1, 0],
    [0, 0],
    [0, 0],
    [1, 1],
    [0, 0]
]

y = ["question", "question", "affirmation", "affirmation", "question", "affirmation"]

model = SVC()
model.fit(X, y)

print(model.predict([[1, 0]]))
print(model.predict([[0, 1]]))
print(model.predict([[0, 0]]))
````

## Exercice 3 — Registre formel ou informel

````python
from sklearn.svm import SVC
#目标：在短消息中使用简单语体特征分类
#每条消息用 3 个特征描述：是否有缩写/是否使用 “您”/是否有表情符号：0 或 1

X = [
    [1, 0, 1],  # "slt :)"
    [1, 0, 0],  # "bjr"
    [0, 1, 0],  # "Pouvez-vous m'aider ?"
    [0, 1, 0],  # "Je vous remercie."
    [1, 0, 1],  # "cc :)"
    [0, 1, 0]   # "Veuillez trouver..."
]

y = ["informel", "informel", "formel", "formel", "informel", "formel"]

model = SVC()
model.fit(X, y)

print(model.predict([[1, 0, 0]]))
print(model.predict([[0, 1, 1]]))
````

## Exercice 4 — Avis positif ou négatif

````python
from sklearn.svm import SVC
#目标：基于简单词汇计数做二分类
#每条评价用两个特征表示：正面词汇数量、负面词汇数量

X = [
    [3, 0],
    [2, 0],
    [0, 3],
    [0, 2],
    [2, 1],
    [1, 2]
]

y = ["positif", "positif", "negatif", "negatif", "positif", "negatif"]

model = SVC()
model.fit(X, y)

print(model.predict([[3, 1]]))
print(model.predict([[0, 4]]))
print(model.predict([[1, 1]]))
````

## Exercice 5 — Passer au texte avec CountVectorizer

````python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.svm import SVC
#目标：学习在 SVM 训练前对文本做自动向量化表示
#CountVectorizer：将每段文本转为词频向量

texts = [
    "bonjour madame je vous remercie",
    "salut ça va",
    "veuillez recevoir mes salutations",
    "coucou merci :)",
    "je vous prie d accepter",
    "hello :)"
]

labels = ["formel", "informel", "formel", "informel", "formel", "informel"]

vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts) #用CountVectorizer 对文本做转换

model = SVC()
model.fit(X, labels) #用转换后的向量训练SVM

#预测下列句子类别：
test = vectorizer.transform(["bonjour", "coucou ça va", "je vous remercie"])
print(model.predict(test))
````
