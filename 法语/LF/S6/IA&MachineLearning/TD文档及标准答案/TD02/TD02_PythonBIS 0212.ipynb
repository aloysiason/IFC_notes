### Exercice 1 -	PARCOURIR UN MOT

````python
# Demander un mot à l'utilisateur
mot = input("Entrez un mot : ")

# 1. Afficher chaque lettre du mot sur une ligne
print("Lettres du mot :")
for lettre in mot:
    print(lettre)

# 2. Compter le nombre de lettres sans utiliser len
compteur = 0  # 初始化字母计数器（用于无len统计数量）
for lettre in mot:
    compteur += 1

print("Nombre de lettres :", compteur)

# 3. Afficher les lettres en majuscules
print("Lettres en majuscules :")
for lettre in mot:
    print(lettre.upper())
````

### Exercice 2 -	Parcourir un texte
https://docs.python.org/fr/3/library/stdtypes.html#text-sequence-type-str

````python
texte = "Python est TOP à 100% ! J'aime coder en 2026 :-D"

# Initialisation des compteurs
nb_majuscules = 0
nb_minuscules = 0
nb_chiffres = 0
nb_espaces = 0
nb_speciaux = 0

# Parcours du texte
for caractere in texte:
    match caractere:
        case c if c.isupper():
            nb_majuscules += 1
        case c if c.islower():
            nb_minuscules += 1
        case c if c.isdigit():
            nb_chiffres += 1
        case " ":
            nb_espaces += 1
        case _:
            nb_speciaux += 1

# Affichage des résultats
print("Nombre de majuscules :", nb_majuscules)
print("Nombre de minuscules :", nb_minuscules)
print("Nombre de chiffres :", nb_chiffres)
print("Nombre d'espaces :", nb_espaces)
print("Nombre de caractères spéciaux :", nb_speciaux)

````

### Exercice 3 -	Fonction analyser_texte

````python
# Définition de la fonction
def analyser_texte(text):
    nb_majuscules = 0
    nb_minuscules = 0
    nb_chiffres = 0
    nb_espaces = 0
    nb_speciaux = 0

    # Parcours du texte
    for caractere in text:
        match caractere:
            case c if c.isupper():
                nb_majuscules += 1
            case c if c.islower():
                nb_minuscules += 1
            case c if c.isdigit():
                nb_chiffres += 1
            case " ":
                nb_espaces += 1
            case _:
                nb_speciaux += 1

    # Affichage des résultats
    print("Nombre de majuscules :", nb_majuscules)
    print("Nombre de minuscules :", nb_minuscules)
    print("Nombre de chiffres :", nb_chiffres)
    print("Nombre d'espaces :", nb_espaces)
    print("Nombre de caractères spéciaux :", nb_speciaux)


# Programme principal
texte_utilisateur = input("Entrez un texte : ")

# Appel de la fonction avec le texte saisi
analyser_texte(texte_utilisateur)
````

### Exercice 4 -	Message caché

````python
# Dictionnaire de substitution
code_secret = {
    "a": "ι",   # iota (grec)
    "e": "κ",   # kappa (grec)
    "m": "μ",   # mu (grec)
    "o": "ω",   # omega (grec)

    "s": "₪",   # symbole (shekel)
    "t": "ת",   # tav (hébreu)
    "r": "ر",   # ra (arabe)
    "h": "ה",   # he (hébreu)
    "y": "ي",   # ya (arabe)

    "x": "@",
    "c": "©",
    "p": "&"
}

def message_cache(texte, code):
    """
    Transforme un texte en message caché en remplaçant certaines lettres
    selon le dictionnaire 'code'. Les autres caractères sont conservés.
    """
    resultat = ""

    # On simplifie : on travaille en minuscules
    texte = texte.lower()

    for caractere in texte:
        if caractere in code:
            resultat += code[caractere]
        else:
            resultat += caractere

    return resultat

# Test
texte_test = "Bonjour, j'aime Python. Je code un message secret."
cache = message_cache(texte_test, code_secret)

print("Texte original :", texte_test)
print("Message caché  :", cache)
````

### Exercice 5 -	Portée des variables

````python
x = 10
print("A) x =", x)  # 顺序1 全局变量x=10

def calculer(x):  # 定义函数（仅加载，不执行内部代码，直到被调用）
    print("B) Dans la fonction (avant modification), x =", x)  # 顺序3 x=3
    x = x + 5
    print("C) Dans la fonction (après modification), x =", x)  # 顺序4 x=8

x = 3
print("D) Avant l'appel de la fonction, x =", x)  # 顺序2

calculer(x)  # 调用函数applle de fonction，执行函数内部代码，此时全局变量x=3

print("E) Après l'appel de la fonction, x =", x)  # 顺序5：修改函数内的局部变量不会影响全局变量 x=3

x = x + 2
print("F) x =", x)  # 顺序6 x=5
````

### Exercice 6 -	Analyse des phrases d’un texte

````python
texte = input("Entrez un texte (avec des phrases séparées par des points) : ")

# 1. Découper le texte en phrases
phrases = texte.split(".")

# Supprimer les phrases vides (si le texte finit par un point)
phrases_nettoyees = []
for phrase in phrases:
    phrase = phrase.strip()  # enlever les espaces au début et à la fin
    if phrase != "":
        phrases_nettoyees.append(phrase)

# 2. Afficher chaque phrase
print("\nPhrases trouvées :")
for phrase in phrases_nettoyees:
    print("-", phrase)

# 3. Nombre de phrases
print("\nNombre de phrases :", len(phrases_nettoyees))

# 4. Analyse de chaque phrase
print("\nAnalyse des phrases :")
for phrase in phrases_nettoyees:
    nb_caracteres = len(phrase)
    mots = phrase.split()
    nb_mots = len(mots)

    print("Phrase :", phrase)
    print("  Nombre de caractères :", nb_caracteres)
    print("  Nombre de mots :", nb_mots)
````

### Exercice 6 -	Nettoyage d’un catalogue d’œuvres

On dispose d’un catalogue d’œuvres littéraires représenté par une liste de tuples de la forme :

`(id, titre, auteur, annee)`

Certaines années peuvent être mal saisies (mauvais type ou valeur incohérente).

```python
catalogue = [("D01", "Du côté de chez Swann",  "Proust",      1913),
             ("D02", "L’Étranger",             "Camus",       "Camus"),
             ("D03", "La Nausée",              "Sartre",      1938),
             ("D04", "Les Fleurs du mal",      "Baudelaire",  " 1857"),
             ("D05", "Le Joueur d’échecs",     "Zweig",       2942)]
```

---

#### 1) Fonction `normaliser_annee(catalogue)`

Écrire une fonction `normaliser_annee` qui parcourt le catalogue et retourne un **nouveau catalogue normalisé**.

Pour chaque entrée :
- si l’année est un entier, elle est conservée ;

- si l’année est une chaîne de caractères :
  - si elle représente un nombre, elle est convertie en entier et le message suivant est affiché :
    ```
    L'année de DXX a été mise sous forme de int
    ```
  - sinon, l’entrée est ignorée et le message suivant est affiché :
    ```
    L'année de DXX est une chaîne de caractères non numérique
    ```
- dans tous les autres cas, l’entrée est ignorée et le message suivant est affiché :
  ```
  L'année de DXX a un type invalide
  ```

---

#### 2) Fonction `valider_annee(catalogue)`

Écrire une fonction `valider_annee` qui prend en entrée un catalogue déjà normalisé et retourne un **nouveau catalogue valide**.

Pour chaque entrée :

- vérifier que l’année est **inférieure ou égale à l’année courante**. Pour obtenir l’année courante, il vous faudra ajouter l’import suivant :
  ```python
  from datetime import date
  ```

- si l’année est future, l’entrée est ignorée et le message suivant est affiché :
  ```
  L'année de DXX dépasse la date actuelle
  ```
- sinon, l’entrée est conservée.



````python
from datetime import date

catalogue = [
    ("D01", "Du côté de chez Swann",  "Proust",      1913),
    ("D02", "L’Étranger",             "Camus",       "Camus"),
    ("D03", "La Nausée",              "Sartre",      1938),
    ("D04", "Les Fleurs du mal",      "Baudelaire",  " 1857"),
    ("D05", "Le Joueur d’échecs",     "Zweig",       2942),
]

def normaliser_annee(catalogue):
    catalogue_normalise = []

    for doc_id, titre, auteur, annee in catalogue:
        if isinstance(annee, int):
            annee_corr = annee

        elif isinstance(annee, str):
            if annee.strip().isdigit():
                print(f"L'année de {doc_id} a été mise sous forme de int")
                annee_corr = int(annee.strip())
            else:
                print(f"L'année de {doc_id} est une chaîne de caractères non numérique")
                continue
        else:
            print(f"L'année de {doc_id} a un type invalide")
            continue

        catalogue_normalise.append((doc_id, titre, auteur, annee_corr))

    return catalogue_normalise


def valider_annee(catalogue):
    catalogue_valide = []
    annee_courante = date.today().year

    for doc_id, titre, auteur, annee in catalogue:
        if annee > annee_courante:
            print(f"L'année de {doc_id} dépasse la date actuelle")
            continue

        catalogue_valide.append((doc_id, titre, auteur, annee))

    return catalogue_valide


catalogue_normalise = normaliser_annee(catalogue)
catalogue_final = valider_annee(catalogue_normalise)

print(catalogue_final)
````
