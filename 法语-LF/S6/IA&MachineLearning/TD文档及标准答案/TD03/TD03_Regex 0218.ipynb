Exercice 1 - MISSION DETOX

```python
import re  #导入 Python 的re模块

txt = "J'ai mangé plein de bonbons, de chocolat, de pizza, de gâteaux et de chips."

#用于替换 “plein de” 之后所有内容的正则表达式
pattern = r"plein de .*"  # r(raw)代表原始字符串(如果不用原始字符串，要表示\，需要在字符串中写两个\\)

replacement = "seulement une salade."

# 使用 re.sub () 函数将匹配到的模式替换为指定的替换字符串
new_txt = re.sub(pattern, replacement, txt)

print("Phrase nouvelle:")
print(new_txt)
print("\nChallenge « stop au sucre » réussi ! ;-) ")
```

Exercice 3 - COMPTER LES NUMEROS PAR PAYS

```python
import re
# 4) 创建国家代码字典
Pays_codes = {
    "France": r"^(\+33|0033)",
    "Argentine": r"^\+54",
    "Royaume-Uni": r"^\+44",
    "USA": r"^\+1",
    "Allemagne": r"^\+49",
    "Espagne": r"^\+34",
    "Italie": r"^\+39"
}

# 引出统计每个国家出现次数的字典并将其初始化=0
compteur = {pays: 0 for pays in Pays_codes}

# 5) 读取txt文件
with open("telephones.txt", "r") as f:
    nums = f.readlines()

# 6)7) 识别每个号码的国际代码并统计
for num in nums:
    num = num.strip()

    for pays, pattern in Pays_codes.items():
         if re.match(pattern, num):
            compteur[pays] += 1
            break
# 8) 显示结果
print("Nombre de numéros par pays :\n")
for pays, nombre in compteur.items():
    print(f"{pays} : {nombre}")
```

Exercice 4 - OPERATION “RATTRAPAGE”

```python
import re
import random #用于生成随机 bonus 分数

#思路：冒号前匹配名字，后匹配分数，冒号周围可能有空格
pattern = r"^(.+?)\s*:\s*(\d+)\s*$"
# 与pattern = r"(\w+)\s*:\s*(\d+)"的对比如下：
# (.+?)允许名字中有-，?限制匹配到第一个冒号就停止
# \s*$代表末尾可能有空格

with open("notes.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        ligne = ligne.strip()
        m = re.match(pattern, ligne)
        if m:
            nom = m.group(1) #提取第一个分组的内容 → 学生名字
            note = int(m.group(2)) #提取第二个分组的内容 → 分数（字符串）

            if note < 10:
                rattrapage = min(20, note + random.randint(5, 10))
               #random.randint()生成一个 5 到 10 之间的随机整数（包含 5 和 10），作为bonus
               #min() 确保分数不超过 20
                print(f"Nom = {nom} | Note = {note} | Rattrapage = {rattrapage}")

            else:
                print(f"Nom = {nom} | Note = {note} ")
```

Exercice 5 - TRANSFORMATION DES DATES

```python
import re

# Dictionnaire des mois
mois = {
    "janvier": "01",
    "février": "02",
    "mars": "03",
    "avril": "04",
    "mai": "05",
    "juin": "06",
    "juillet": "07",
    "août": "08",
    "septembre": "09",
    "octobre": "10",
    "novembre": "11",
    "décembre": "12"
}


pattern = r"^\s*\w+\s+(\d{1,2})\s+([^\s]+)\s+(\d{4})\s*$"
 #\s*匹配行首可能的空格；\w+匹配星期名称；\s+匹配空格；\d{1,2})匹配日；(\d{4})	匹配年
 #[^\s] 表示“不是空白字符的任意字符”，匹配月（因为月份有重音符号，所以无法使用/w）
with open("dates.txt", "r", encoding="utf-8") as f:
    for ligne in f:
        ligne = ligne.strip()

        m = re.match(pattern, ligne)
        if not m:
            print("Ligne ignorée (format invalide) :", ligne)
            continue #如果匹配失败（格式不对），打印提示并跳过这一行

  #用 .group(1) .group(2) .group(3) 提取三个分组的内容
        jour = m.group(1)
        nom_mois = m.group(2).lower()  # .lower() 将月份名转为小写，与字典的键一致
        annee = m.group(3)

        if nom_mois not in mois:
            print("Mois inconnu :", nom_mois, "dans", ligne)
            continue #检查月份名是否在字典中，如果不在，打印提示并跳过

        jj = jour.zfill(2) #将日数字转为两位，不足两位前面补零
        mm = mois[nom_mois] #从字典中获取月份对应的两位数字

  #按题目要求的格式输出：原行 -> 转换后的 jj/mm/aaaa
        print(f"{ligne} ----> {jj}/{mm}/{annee}")
```
